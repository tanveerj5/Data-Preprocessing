importing the libraries

```python
import numpy as np # arrays
import matplotlib.pyplot as plt # graph 
import pandas as pd # importing datasets and works with matrix
```

importing the datasets

```python
datasets = pd.read_csv('Data.csv') # Dataframe
x = datasets.iloc[:, :-1].values # all rows except last column and .values convert into arrays of 2d
y = datasets.iloc[:, -1].values # all rows only last column
```

```python
x # independent variables
```

```python
y # dependent variables
```

Take Care of Missing Data

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean') # imputer is for central tendency has mean, median and mode
imputer.fit(x[:, 1:3]) # we apply the changes of mean for interval data(age, salary) float datasets
x[:, 1:3] = imputer.transform(x[:, 1:3]) # we transform the matrix to apply the changes and return result to both age, salary columns
```

```python
x
```

Encoding categorical data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# one-hot encoding

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
x = np.array(ct.fit_transform(x))
```

```python
x
```

```python
# Label Encoder
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(y)
```

```python
y
```

splitting datasets into Train and Test sets

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler() # Magnitude and Direction for normal distrbution
x_train[:, 3:] = sc.fit_transform(x_train[:, 3:])
x_test[:, 3:] = sc.transform(x_test[:, 3:])
```
